```python
# -----------------------------
# Hospital Management System
# -----------------------------

class LoginSystem:
    def __init__(self, username, password):
        self.username = username
        self.password = password

    def login(self):
        print(f"{self.username} logged in successfully.")

    def logout(self):
        print(f"{self.username} logged out.")


# -----------------------------
class Admin:
    def __init__(self, admin_id, name):
        self.admin_id = admin_id
        self.name = name

    def add_staff(self):
        print("Staff Added Successfully.")

    def remove_staff(self):
        print("Staff Removed Successfully.")

    def generate_report(self):
        print("Hospital Report Generated.")


# -----------------------------
class Patient:
    def __init__(self, patient_id, name, age, gender):
        self.patient_id = patient_id
        self.name = name
        self.age = age
        self.gender = gender

    def admit(self):
        print(f"{self.name} admitted.")

    def discharge(self):
        print(f"{self.name} discharged.")


# -----------------------------
class Doctor:
    def __init__(self, doctor_id, name, specialization):
        self.doctor_id = doctor_id
        self.name = name
        self.specialization = specialization

    def diagnose(self):
        print(f"Dr. {self.name} diagnosed the patient.")

    def prescribe(self):
        print("Prescription Generated.")


# -----------------------------
class Nurse:
    def __init__(self, nurse_id, name):
        self.nurse_id = nurse_id
        self.name = name

    def check_vitals(self):
        print("Vitals Checked.")

    def assist_doctor(self):
        print("Doctor Assisted.")


# -----------------------------
class Receptionist:
    def register_patient(self):
        print("Patient Registered.")

    def book_appointment(self):
        print("Appointment Booked.")


# -----------------------------
class Appointment:
    def __init__(self, appointment_id, date, time):
        self.appointment_id = appointment_id
        self.date = date
        self.time = time

    def schedule(self):
        print("Appointment Scheduled.")

    def cancel(self):
        print("Appointment Cancelled.")


# -----------------------------
class MedicalHistory:
    def __init__(self, disease, allergy):
        self.disease = disease
        self.allergy = allergy

    def show_history(self):
        print("Disease:", self.disease)
        print("Allergy:", self.allergy)


# -----------------------------
class Prescription:
    def __init__(self, medicine, dosage):
        self.medicine = medicine
        self.dosage = dosage

    def print_prescription(self):
        print(f"Medicine: {self.medicine}")
        print(f"Dosage: {self.dosage}")


# -----------------------------
class Billing:
    def __init__(self, amount):
        self.amount = amount

    def generate_bill(self):
        print(f"Bill Amount: ₹{self.amount}")


# -----------------------------
class Payment:
    def __init__(self, mode):
        self.mode = mode

    def pay(self):
        print(f"Payment Done by {self.mode}")


# -----------------------------
class Room:
    def __init__(self, room_no):
        self.room_no = room_no

    def assign_room(self):
        print(f"Room {self.room_no} Assigned.")


# -----------------------------
class Ward:
    def allocate_bed(self):
        print("Bed Allocated.")


# -----------------------------
class ICU:
    def admit_patient(self):
        print("Patient Shifted to ICU.")


# -----------------------------
class Pharmacy:
    def issue_medicine(self):
        print("Medicine Issued.")


# -----------------------------
class Medicine:
    def __init__(self, name, price):
        self.name = name
        self.price = price

    def display(self):
        print(self.name, self.price)


# -----------------------------
class Laboratory:
    def conduct_test(self):
        print("Lab Test Completed.")


# -----------------------------
class LabReport:
    def print_report(self):
        print("Lab Report Printed.")


# -----------------------------
class Ambulance:
    def book(self):
        print("Ambulance Booked.")


# -----------------------------
class Insurance:
    def verify(self):
        print("Insurance Verified.")


# -----------------------------
class Attendance:
    def mark(self):
        print("Attendance Marked.")


# -----------------------------
class Salary:
    def __init__(self, salary):
        self.salary = salary

    def show_salary(self):
        print("Salary:", self.salary)


# -----------------------------
class Feedback:
    def submit(self):
        print("Feedback Submitted.")


# =============================
# Main Program
# =============================

login = LoginSystem("admin", "1234")
login.login()

patient = Patient(101, "chirag", 25, "Male")
patient.admit()

doctor = Doctor(1, "Raj", "Cardiologist")
doctor.diagnose()
doctor.prescribe()

appointment = Appointment(201, "01-08-2026", "10:30 AM")
appointment.schedule()

bill = Billing(4000)
bill.generate_bill()

payment = Payment("UPI")
payment.pay()

feedback = Feedback()
feedback.submit()

login.logout()
```
